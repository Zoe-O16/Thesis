# Intercoder reliability sample

Merging coder 1, coder 2, LLM coder and the model's own topic assignments for the 200 sample reviews from 2.7.

```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score

# Coder paths
CODER1 = "data/coder_1.csv"
CODER2 = "data/coder_2.csv"
CODERLLM = "data/coder_llm.csv"
```

```python
sample = pd.read_csv("data/coding_sample_200.csv")
sample = sample.drop(columns=["topic_number"])
sample.head()
```

```python
model_df = pd.read_csv("outputs_rlda_codebook/codebook_dominant_topic_assignments.csv")
model_df = model_df[["doc_id", "dominant_topic"]].rename(columns={"doc_id": "review_id", "dominant_topic": "model_assignment"})

sample = sample.merge(model_df, on="review_id", how="left")
sample.head()
```

```python
coder1_df = pd.read_csv(CODER1)
coder1_df = coder1_df[["review_id", "topic_number"]].rename(columns={"topic_number": "coder_1"})
coder1_df["coder_1"] = coder1_df["coder_1"].astype("Int64")

sample = sample.merge(coder1_df, on="review_id", how="left")
sample.head()
```

```python
coder2_df = pd.read_csv(CODER2)
coder2_df = coder2_df[["review_id", "topic_number"]].rename(columns={"topic_number": "coder_2"})
coder2_df["coder_2"] = coder2_df["coder_2"].astype("Int64")


sample = sample.merge(coder2_df, on="review_id", how="left")
sample.head()
```

```python
llm_df = pd.read_csv(CODERLLM)
llm_df = llm_df[["review_id", "topic_number"]].rename(columns={"topic_number": "llm_coder"})

sample = sample.merge(llm_df, on="review_id", how="left")
sample.head()
```

```python
sample = sample[["review_id", "year", "season", "designer", "author", "city", "date", "review",
                  "coder_1", "coder_2", "llm_coder", "model_assignment"]]

sample.to_csv("data/coding_sample_200_intercoder.csv", index=False)
print(len(sample))
sample.head()
```

```python
raters = ["coder_1", "coder_2", "llm_coder", "model_assignment"]

for i in range(len(raters)):
    for j in range(i + 1, len(raters)):
        r1 = raters[i]
        r2 = raters[j]
        pair = sample[[r1, r2]].dropna()
        kappa = cohen_kappa_score(pair[r1], pair[r2])
        print(r1, "vs", r2, "-> kappa =", round(kappa, 3), "(n =", len(pair), ")")
```
